```python
#TUGAS D3
import cv2
import matplotlib.pyplot as plt
import numpy as np

# Load gambar KTM
path_ktm = (
    "/content/drive/MyDrive/SEMESTER 5/Citra/KTM (Kartu Tanda Mahasiswa).jpg")
img_bgr = cv2.imread(path_ktm)
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

# 1. Eksplorasi Channel Warna (R, G, B)
R, G, B = cv2.split(img_rgb)

# 2. Resizing & Flip (Flip Vertikal & Horizontal / Code 0 & -1)
img_resized = cv2.resize(img_rgb, (400, 250))
img_flip_v = cv2.flip(img_resized, 0)  # Vertical
img_flip_hv = cv2.flip(img_resized, -1)  # Both

# 3. Menggambar Geometri di atas Gambar KTM
img_geo = img_resized.copy()
cv2.rectangle(img_geo, (50, 50), (150, 150), (255, 0, 0), 3)  # Kotak Merah di area foto
cv2.circle(img_geo, (300, 100), 30, (0, 255, 0), -1)  # Lingkaran Hijau Terisi (Filled)
cv2.putText(img_geo, "VERIFIED", (180, 200), cv2.FONT_HERSHEY_SIMPLEX, 0.8, (255, 255, 0), 2,)

# Visualisasi Hasil
fig, axes = plt.subplots(2, 3, figsize=(15, 8))

axes[0, 0].imshow(R, cmap="Reds")
axes[0, 0].set_title("Channel Red")
axes[0, 1].imshow(G, cmap="Greens")
axes[0, 1].set_title("Channel Green")
axes[0, 2].imshow(B, cmap="Blues")
axes[0, 2].set_title("Channel Blue")

axes[1, 0].imshow(img_flip_v)
axes[1, 0].set_title("Flip Vertikal")
axes[1, 1].imshow(img_flip_hv)
axes[1, 1].set_title("Flip Both (H+V)")
axes[1, 2].imshow(img_geo)
axes[1, 2].set_title("Geometri pada KTM")

for ax in axes.ravel():
  ax.axis("off")
plt.tight_layout()
plt.show()
```
