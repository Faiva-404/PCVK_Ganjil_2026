```python
#D2
import numpy as np
import pandas as pd
import cv2 as cv
from google.colab.patches import cv2_imshow # for image display
from skimage import io
from skimage import transform
from PIL import Image
import matplotlib.pylab as plt
```

```python
# Membuat list untuk menyimpan url dari beberapa image
urls = ["https://iiif.lib.ncsu.edu/iiif/0052574/full/800,/0/default.jpg",
        "https://iiif.lib.ncsu.edu/iiif/0016007/full/800,/0/default.jpg"]
# baca dan tampilkan image
for url in urls:
  image = io.imread(url)
  image = cv.resize(image, (0,0), fx=0.5, fy=0.5)
  image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
  final_frame = cv.hconcat((image, image_2))
  cv2_imshow(final_frame)

  print("\n")
```

```python
#Melihat ukuran file image
tinggi = image_2.shape[0]
lebar = image_2.shape[1]
print("resolusi image: tinggi x lebar = ", tinggi, " x ", lebar)
cv2_imshow(image_2)
```

```python
#akses pixel dengan garis horizontal berwarna putih
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

#garis horizontal ditengah image
for y in range(lebar):
  image_3[int((tinggi)/2),y] = [255,255,255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```

```python
#TUGAS PRAKTIKUM D3
#2. Menampilkan Citra Channel Red-Blue dan Green-Blue
import cv2
import matplotlib.pyplot as plt
from skimage import io

url = "/content/drive/MyDrive/SEMESTER 5/Citra/KTM (Kartu Tanda Mahasiswa).jpg"
img_rgb = io.imread(url)

# Channel Red-Blue (Green dinolkan)
img_rb = img_rgb.copy()
img_rb[:, :, 1] = 0

# Channel Green-Blue (Red dinolkan)
img_gb = img_rgb.copy()
img_gb[:, :, 0] = 0

# Visualisasi
fig, axes = plt.subplots(1, 2, figsize=(10, 5))
axes[0].imshow(img_rb)
axes[0].set_title("Channel Red-Blue")
axes[0].axis("off")

axes[1].imshow(img_gb)
axes[1].set_title("Channel Green-Blue")
axes[1].axis("off")

plt.show()
```

```python
#7. Rectangle & Circle pada Area Wajah

#up foto
path_foto = "/content/drive/MyDrive/SEMESTER 5/Citra/fotoku.jpeg"
img_bgr = cv2.imread(path_foto)

#Cek apakah file lokal berhasil dibaca
if img_bgr is None:
  print("File lokal tidak ditemukan!")
  print("Pastikan nama file dan lokasi di folder /content/ sudah benar.")
else:
  # Konversi warna ke RGB (visualisasi) dan Gray (deteksi)
  img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
  img_gray = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2GRAY)

  # 3. Load classifier deteksi wajah
  face_cascade = cv2.CascadeClassifier(
      cv2.data.haarcascades + "haarcascade_frontalface_default.xml"
  )
  faces = face_cascade.detectMultiScale(img_gray, scaleFactor=1.1, minNeighbors=8, minSize=(100, 100))

  img_result = img_rgb.copy()

  # 4. Gambar Rectangle dan Circle pada area wajah
  for x, y, w, h in faces:
    center = (x + w // 2, y + h // 2)
    radius = w // 2

    # Rectangle (Kotak Kuning) & Circle (Lingkaran Cyan)
    cv2.rectangle(img_result, (x, y), (x + w, y + h), (255, 255, 0), 3)
    cv2.circle(img_result, center, radius, (0, 255, 255), 2)

  # 5. Tampilkan hasil
  plt.figure(figsize=(6, 6))
  plt.imshow(img_result)
  plt.title(f"Deteksi Wajah Lokal (Ditemukan: {len(faces)})")
  plt.axis("off")
  plt.show()
```
